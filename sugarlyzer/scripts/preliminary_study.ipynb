```python
import os
import json

candidate_configs = {f"{x}.config" for x in range(0, 100)}
results_dir = "../results_1000"

for root, dirs, files in os.walk(results_dir):
    for file in files:
        if file.endswith(".json") and "product" in root:
            num_results_okay = 0
            with open(os.path.join(root, file), 'r') as f:
                data = json.load(f)
            for res in data:
                intersection = set(res['configFiles']).intersection(candidate_configs)
                if len(intersection) > 0:
                    num_results_okay += 1
            print(f"{os.path.join(root, file)}: {num_results_okay}/{len(data)} results with config in candidate configs")
```
