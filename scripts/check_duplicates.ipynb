```python
import os
import json
from csv import DictReader
from typing import Dict
from taxonomy import *

bugs_directory = "../../results/ESEM_26/bug_reports"


for p in ['PRODUCT', 'TRANSFORMATION', 'FAMILY']:
    bugs_to_ids: Dict[tuple, list] = {}
    for root, dirs, files in os.walk(bugs_directory):
        for file in files:
            if file.endswith(".json") and p in file and '#' not in file:
                with open(os.path.join(root, file), "r") as f:
                    bug_report = json.load(f)
                if p == 'PRODUCT':
                    key = (bn := os.path.basename(bug_report['originalAlarm']['fileLocation']),
                           bug_report['originalAlarm']['line'],
                           bug_report['originalAlarm']['description'],
                           bug_report['presenceCondition'])
                    if bug_report['id'] in ['PRODUCT_57', 'PRODUCT_200']:
                        print(key)
                    if key not in bugs_to_ids:
                        bugs_to_ids[key] = []
                    bugs_to_ids[key].append(bug_report['id'])
                elif p == 'TRANSFORMATION':
                    key = (bn := os.path.basename(bug_report['originalAlarm']['fileLocation']),
                           tuple(bug_report['lineInputFile']),
                           bug_report['sanitizedDescription'],
                           bug_report['presenceCondition'])
                    if key not in bugs_to_ids:
                        bugs_to_ids[key] = []
                    bugs_to_ids[key].append(bug_report['id'])
                elif p == 'FAMILY':
                    key = (os.path.basename(bug_report['input_file']),
                           bug_report['line_number'],
                           bug_report['err'])
                    if key not in bugs_to_ids:
                        bugs_to_ids[key] = []
                    bugs_to_ids[key].append(bug_report['id'])
    for key, ids in bugs_to_ids.items():
        if len(ids) > 1:
            print(f"Key: {key} has {len(ids)} matching bugs: {ids}")
```
